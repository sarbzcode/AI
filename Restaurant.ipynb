```python
import numpy as np
import pandas as pd
```

```python
Restaurant = pd.read_csv("/content/restaurant.csv")
```

```python
Restaurant
```

```python
from sklearn.preprocessing import LabelEncoder
Le = LabelEncoder()

Restaurant['Alt'] = Le.fit_transform(Restaurant['Alt'])
Restaurant['Bar'] = Le.fit_transform(Restaurant['Bar'])
Restaurant['Fri'] = Le.fit_transform(Restaurant['Fri'])
Restaurant['Hun'] = Le.fit_transform(Restaurant['Hun'])
Restaurant['Pat'] = Le.fit_transform(Restaurant['Pat'])
Restaurant['Price'] = Le.fit_transform(Restaurant['Price'])
Restaurant['Rain'] = Le.fit_transform(Restaurant['Rain'])
Restaurant['Res'] = Le.fit_transform(Restaurant['Res'])
Restaurant['Type'] = Le.fit_transform(Restaurant['Type'])
Restaurant['Est'] = Le.fit_transform(Restaurant['Est'])
Restaurant['Wait'] = Le.fit_transform(Restaurant['Wait'])
```

```python
Restaurant
```

```python
y = Restaurant['Wait']
X = Restaurant.drop(['Wait'],axis=1)
```

```python
from sklearn import tree
clf = tree.DecisionTreeClassifier(criterion='entropy')
clf = clf.fit(X, y)
```

```python
tree.plot_tree(clf)
```

```python
import graphviz
dot_data = tree.export_graphviz(clf, out_file=None)
graph = graphviz.Source(dot_data)
graph
```

```python
X_pred = clf.predict(X)
```

```python
X_pred == y
```
